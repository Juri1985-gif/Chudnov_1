```python
class Chocolate:
  def __init__(self, price: float, additive: str = None):
    if price <= 0:
      raise ValueError("Цена меньше нуля!")

    self.price = price
    self.additive = additive

  def show_my_chocolate(self):
    if self.additive:
      print(f"Шоколадный батончик, добавка: {self.additive}")
    else:
      print("Обычный шоколадный батончик")

  def __str__(self):
    return f'Шоколадный батончик, {self.price} рублей'


print("Пример 1:")
Chocolate_1 = Chocolate(100)
Chocolate_1.show_my_chocolate()
print(Chocolate_1)

print("Пример 2:")
Chocolate_2 = Chocolate(50, "молоко")
Chocolate_2.show_my_chocolate()
print(Chocolate_2)

print("Пример 3:")
Chocolate_3 = Chocolate(-10, "изюм")
Chocolate_3.show_mt_chocolate()
print(Chocolate_3)
```

**Выводы:**

Для оптимизации типы столбцов price, bedrooms, bathrooms, stories и parking можно изменить с int64 на int8. Тип столбца area на int16.
Столбцы mainroad, guestroom, basement, hotwaterheating, airconditioning, prefarea и furnishingstatus содержат строковые данные; для уменьшения занимаемой памяти их следует преобразовать из типа object в category.


```python
df_housing.describe()
```

```python
df_housing = df_housing.fillna(0)
df_housing['price'] = df_housing['price'].astype(np.int8)
df_housing['area'] = df_housing['area'].astype(np.int16)
df_housing['bedrooms'] = df_housing['bedrooms'].astype(np.int8)
df_housing['bathrooms'] = df_housing['bathrooms'].astype(np.int8)
df_housing['stories'] = df_housing['stories'].astype(np.int8)
df_housing['parking'] = df_housing['parking'].astype(np.int8)
```

```python
df_housing.info()
```

```python
def extract_first_name(name):
    return name.split(',')[1].split('.')[1].strip().split(' ')[0]
```

```python
df['FirstName'] = df['Name'].apply(extract_first_name)
df[df['Sex'] == 'male']['FirstName'].mode()[0]
df[df['Sex'] == 'female']['FirstName'].mode()[0]
```

```python
print("\n--- Популярные имена по классам ---")
for pclass in sorted(df['Pclass'].unique()):
    class_data = df[df['Pclass'] == pclass]
    m_name = class_data[class_data['Sex'] == 'male']['FirstName'].mode()
    m_res = m_name[0] if not m_name.empty else "N/A"
    print(f"Класс {pclass}: Мужское - {m_res}")
```

```python
print("\n--- Популярные имена по классам ---")
for pclass in sorted(df['Pclass'].unique()):
    class_data = df[df['Pclass'] == pclass]
    f_name = class_data[class_data['Sex'] == 'female']['FirstName'].mode()
    f_res = f_name[0] if not f_name.empty else "N/A"
    print(f"Класс {pclass}: Женское - {f_res}")
```
